## Datos sin sentido
Al ejecutar `describe()` aparecen algunos valores que a simple vista no tienen sentido, por ejemplo,se puede apreciar que en el dataset la edad maxima de una persona es de 144, lo cual es practicamente imposible, si buscamos las edades que sobre pasan los 100 años vemos que existe más de uno con este posible error de escritura, ya que si nos fijamos la duración que tienen trabajando no supera los 13 años de ninguno.

```python
df[df['person_age']>100]
```

De La misma forma vemos que el valor máximo de años trabajando arrojaba como valor 123, lo cual también es imposible. En este caso solo fueron dos registros que presentaban esta anomalía. Se proceden a eliminar los valores con inconsistencias.

```python
df[df['person_emp_length']>60]
```

```python
df = df[df['person_age']<100]
df = df[df['person_emp_length']<60]
```

```python
df.isnull().sum()
```

```python
df[df['loan_int_rate'].isnull()]
```

# Creación de una columna temporal con al mediana que le toca a cada fila

```python
mediananas_por_grado = df.groupby('loan_grade')['loan_int_rate'].median()
mediananas_por_grado
```

```python
# Una vez que tengamos las medianas de cada Grado de Préstamo, crearemos una columna temporal
df['mediana_grupo'] = df['loan_grade'].map(mediananas_por_grado)
```

```python
df[['loan_grade', 'loan_int_rate', 'mediana_grupo']].head(10)
```

```python
# Rellenamos los nulos con los valores de la columna 'mediana_grupo'01_data_cleaning.ipynb
df['loan_int_rate'] = df['loan_int_rate'].fillna(df['mediana_grupo'])
```

```python
df
```

```python
df = df.drop(columns=['mediana_grupo'])
```

```python
# Comprobamos si existen valores nulos
df.isnull().sum()
```

```python
df['person_id'] = range(1, len(df) + 1)
df
```
